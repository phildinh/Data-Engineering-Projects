# Transformations

### Calendar

```python
from pyspark.sql.functions import *

df_cal = df_cal.withColumn('Month', month(col('Date')))\
                .withColumn('Year', year(col('Date')))

df_cal.display()
```

```python
df_cal.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Calendar")\
            .save()
```

### Customers

```python
df_cus = df_cus.withColumn("FullName", concat_ws(' ', col('Prefix'), col('FirstName'), col('LastName')))
```

```python
df_cus.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Customers")\
            .save()
```

### Sub Catgories

```python
df_subcat.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Subcategories")\
            .save()
```

### Products


```python
df_pro = df_pro.withColumn('ProductSKU', split(col('ProductSKU'), '-')[0])\
                .withColumn('ProductName', split(col('ProductName'), ' ')[0])

df_pro.display()
```

```python
df_pro.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Products")\
            .save()
```

### Returns

```python
df_return.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Returns")\
            .save()
```

### Territories

```python
df_ter.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Territories")\
            .save()
```

### Sales

```python
df_sale.display()
```

```python
df_sale = df_sale.withColumn('StockDate', to_timestamp('StockDate', 'MM/dd/yyyy'))\
        .withColumn('OrderNumber', regexp_replace(col('OrderNumber'),'S','T'))\
        .withColumn('multiply', col('OrderLineItem')*col('OrderQuantity'))
```

```python
df_sale.write.format('parquet')\
            .mode('append')\
            .option("path","abfss://silver@azureprojectphil.dfs.core.windows.net/AdventureWorks_Sales")\
            .save()
```
