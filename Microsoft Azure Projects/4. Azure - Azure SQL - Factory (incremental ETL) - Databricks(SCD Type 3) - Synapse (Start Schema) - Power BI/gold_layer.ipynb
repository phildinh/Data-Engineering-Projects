# CREATING DIMENSION MODEL


```python
%sql
SELECT DISTINCT(Model_ID) AS Model_ID, model_category
FROM parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
```

```python
df_src = spark.sql('''    
    SELECT 
    DISTINCT(Model_ID) AS Model_ID,
    model_category,
    current_date() AS effective_date,
    DATE('3000-01-01') AS expire_date,
    TRUE AS status
    FROM parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
    ''').withColumn("dim_model_key", monotonically_increasing_id())  # surrogate key

df_src.display()
```

```python
df_src.write.format('delta')\
            .mode('overwrite')\
            .option('path','abfss://gold@azureprojectphil.dfs.core.windows.net/dim_model')\
            .save()
```

```python
df_branch = spark.sql('''
        SELECT DISTINCT(Branch_ID) AS Branch_ID, 
        BranchName,
        current_date() AS effective_date,
        DATE('3000-01-01') AS expire_date,
        TRUE AS status
        FROM parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
        ''').withColumn("dim_branch_key", monotonically_increasing_id())  # surrogate key   
```

```python
df_branch.write.format('delta')\
            .mode('overwrite')\
            .option('path','abfss://gold@azureprojectphil.dfs.core.windows.net/dim_branch')\
            .save()
```

```python
df_dealer = spark.sql('''
        SELECT DISTINCT(Dealer_ID) AS Dealer_ID, 
        DealderName,
        current_date() AS effective_date,
        DATE('3000-01-01') AS expire_date,
        TRUE AS status
        FROM parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
        ''').withColumn("dim_dealer_key", monotonically_increasing_id())  # surrogate key
```

```python
df_dealer.write.format('delta')\
            .mode('overwrite')\
            .option('path','abfss://gold@azureprojectphil.dfs.core.windows.net/dim_dealer')\
            .save()
```

```python
df_date = spark.sql('''
        SELECT DISTINCT(Date_ID) AS Date_ID, 
        Day,
        Month,
        Year
        FROM parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
        ''').withColumn("dim_date_key", monotonically_increasing_id())  # surrogate key
```

```python
df_date.display()
```

```python
df_date.write.format('delta')\
            .mode('overwrite')\
            .option('path','abfss://gold@azureprojectphil.dfs.core.windows.net/dim_date')\
            .save()
```

# Createing Fact Table


```python
df = spark.sql('''
        SELECT * FROM
        parquet.`abfss://silver@azureprojectphil.dfs.core.windows.net/carsales`
               ''')
```

```python
df.display()
```

```python
df_fact = df.join(df_branch, df.Branch_ID == df_branch.Branch_ID,how='left')\
            .join(df_dealer, df.Dealer_ID == df_dealer.Dealer_ID, how='left')\
            .join(df_src, df.Model_ID == df_src.Model_ID, how='left')\
            .join(df_date, df.Date_ID == df_date.Date_ID, how='left')\
            .select('dim_branch_key', 'dim_dealer_key', 'dim_date_key','dim_model_key','Revenue', 'Units_Sold','RevPerUnit')
```

```python
df_fact.display()
```

```python
df_fact.write.format('delta')\
            .mode('overwrite')\
            .option('path','abfss://gold@azureprojectphil.dfs.core.windows.net/fact_sales')\
            .save()
```
