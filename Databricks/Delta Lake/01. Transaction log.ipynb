# Understand Delta Lake Transaction Log
Understand the cloud storage directory structure behind delta lak tables

### 0. Create demo catalog

```sql
CREATE CATALOG IF NOT EXISTS demo
MANAGED LOCATION 'abfss://demo@philproject1.dfs.core.windows.net/'
```

### 1. Create a new schema under the demo catalog for this section of the course (delta_lake)

```sql
CREATE SCHEMA IF NOT EXISTS demo.delta_lake
  MANAGED LOCATION 'abfss://demo@philproject1.dfs.core.windows.net/delta_lake'
```

### 2. Create a Delta Lake Table

```sql
CREATE TABLE IF NOT EXISTS demo.delta_lake.companies
(company_name STRING,
founded_name DATE,
country STRING);
```

```sql
DESC EXTENDED demo.delta_lake.companies
```
