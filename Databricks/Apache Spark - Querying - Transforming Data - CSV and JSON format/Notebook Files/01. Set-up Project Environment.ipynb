# Create the catalog - gizmobox

```sql
%sql
SHOW CATALOGS;
```

```sql
%sql
CREATE CATALOG  IF NOT EXISTS gizmobox
    MANAGED LOCATION 'abfss://gizmobox@philproject1.dfs.core.windows.net/'
    COMMENT 'This is the catalog for the Gizmobox Data Lakehouse'
```

```sql
USE CATALOG gizmobox;

CREATE SCHEMA IF NOT EXISTS  landing
   MANAGED LOCATION 'abfss://gizmobox@philproject1.dfs.core.windows.net/landing';
CREATE SCHEMA IF NOT EXISTS  bronze
   MANAGED LOCATION 'abfss://gizmobox@philproject1.dfs.core.windows.net/bronze';
CREATE SCHEMA IF NOT EXISTS  silver
   MANAGED LOCATION 'abfss://gizmobox@philproject1.dfs.core.windows.net/silver';
CREATE SCHEMA IF NOT EXISTS  gold
   MANAGED LOCATION 'abfss://gizmobox@philproject1.dfs.core.windows.net/gold';
```
