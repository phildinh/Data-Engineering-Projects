# Set-up the project environment for CircuitBox Data Lakehouse
1. Create external location - philproject1-circuitbox
2. Create catalog - circuitbox
3. Create Schemas
  - landing
  - lakehouse
4. Create volume - operational_data

```sql
-- 1. Create external location
CREATE EXTERNAL LOCATION IF NOT EXISTS philproject1_circuitbox
  URL 'abfss://circuitbox@philproject1.dfs.core.windows.net/'
  WITH (STORAGE CREDENTIAL circuitbox)
  COMMENT 'External location for the circuitbox data lakehouse';
```

```sql
%fs ls 'abfss://circuitbox@philproject1.dfs.core.windows.net'
```

```sql
-- 2.Create catalog
CREATE CATALOG IF NOT EXISTS circuitbox
  MANAGED LOCATION 'abfss://circuitbox@philproject1.dfs.core.windows.net/'
  COMMENT 'catalog for the circuit box data lakehouse';
```

```sql
SHOW CATALOGS;
```

```sql

-- 3. Create schema
USE CATALOG circuitbox;

CREATE SCHEMA IF NOT EXISTS landing
  MANAGED LOCATION 'abfss://circuitbox@philproject1.dfs.core.windows.net/landing'
  COMMENT 'landing schema for the circuit box data lakehouse';

CREATE SCHEMA IF NOT EXISTS lakehouse
  MANAGED LOCATION 'abfss://circuitbox@philproject1.dfs.core.windows.net/lakehouse'
  COMMENT 'lakehouse schema for the circuit box data lakehouse';
```

```sql
SHOW SCHEMAS;
```

```sql
DROP VOLUME IF EXISTS operational_data
```

```sql
-- 4. Create Volume
USE CATALOG circuitbox;
USE SCHEMA landing;

CREATE EXTERNAL VOLUME IF NOT EXISTS operational_data
  LOCATION 'abfss://circuitbox@philproject1.dfs.core.windows.net/landing/operational_data/'
  COMMENT 'volume for the circuit box data lakehouse'
```

```sql
%fs ls /Volumes/circuitbox/landing/operational_data
```
